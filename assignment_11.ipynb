```python

gamma = 0.9
theta = 0.001

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 100)
    }
}

values = {"S0": 0.0, "S1": 0.0, "G": 0.0}

policy = {
    "S0": "Up",
    "S1": "Up"
}

print("INITIAL POLICY")
print(policy)

print("\nTRANSITION TABLE")
for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

iteration = 0

while True:
    iteration += 1

    print("\nPOLICY ITERATION", iteration)

    # Policy Evaluation
    evaluation_step = 0

    while True:
        new_values = values.copy()

        for state in states:
            action = policy[state]
            next_state, reward = transitions[state][action]

            new_values[state] = (
                reward + gamma * values[next_state]
            )

        delta = max(
            abs(new_values[state] - values[state])
            for state in states
        )

        values = new_values
        evaluation_step += 1

        print(
            f"Evaluation {evaluation_step}: "
            f"S0 = {values['S0']:.4f}, "
            f"S1 = {values['S1']:.4f}, "
            f"Delta = {delta:.4f}"
        )

        if delta < theta:
            break

    print("\nPOLICY EVALUATION COMPLETED")
    print("State Values:", values)

    # Policy Improvement
    old_policy = policy.copy()
    q_table = {}

    for state in states:
        q_table[state] = {}

        for action in actions:
            next_state, reward = transitions[state][action]

            q = reward + gamma * values[next_state]
            q_table[state][action] = q

    for state in states:
        best_value = max(q_table[state].values())

        best_actions = [
            action for action in actions
            if abs(q_table[state][action] - best_value) < 1e-9
        ]

        policy[state] = best_actions[0]

    print("\nPOLICY IMPROVEMENT")
    print("Old Policy:", old_policy)
    print("New Policy:", policy)

    # Policy Stability
    if old_policy == policy:
        print("\nPOLICY IS STABLE")
        break

# Final Results
print("\nFINAL OPTIMAL STATE VALUES")

for state in states:
    print(f"V*({state}) = {values[state]:.4f}")

print(f"V*(G) = {values['G']:.4f}")

# Final Action Values
print("\nFINAL ACTION VALUE TABLE")

q_table = {}

for state in states:
    q_table[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]

        q = reward + gamma * values[next_state]
        q_table[state][action] = q

        print(f"Q*({state}, {action}) = {q:.4f}")

# Final Optimal Policy
print("\nFINAL OPTIMAL POLICY")

for state in states:
    print(state, "=", policy[state])

# Bellman Verification
print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in states:
    max_q = max(q_table[state].values())

    verified = abs(values[state] - max_q) < 1e-9

    print(
        state,
        "V* =", round(values[state], 4),
        "Max Q* =", round(max_q, 4),
        "Verified =", verified
    )

print("\nOPTIMAL PATH: S0 -> S1 -> G")
```
