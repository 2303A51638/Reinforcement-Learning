```python
# Assignment-1
import numpy as np

gamma = 0.9

A = np.array([[1, -0.9 * 0.9],[-0.1 * 0.9, 1]], dtype=float)

b = np.array([-0.9 + 0.1 * (-1),0.9 * 100 + 0.1 * (-1)], dtype=float)

V0, V1 = np.linalg.solve(A, b)

Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

print("Assignment 1: Algebraic Bellman Solution")
print(f"V(S0)       = {V0:.2f}")
print(f"V(S1)       = {V1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right)= {Q_S0_Right:.2f}")
print(f"Q(S1, Right)= {Q_S1_Right:.2f}")
print(f"Q(S1, Up)   = {Q_S1_Up:.2f}")

print("\nVerification")
print(f"V(S0) = 0.9*Q(S0, Down) + 0.1*Q(S0, Right) = {0.9*Q_S0_Down + 0.1*Q_S0_Right:.2f}")
print(f"V(S1) = 0.9*Q(S1, Right) + 0.1*Q(S1, Up) = {0.9*Q_S1_Right + 0.1*Q_S1_Up:.2f}")
```

```python
#Assignment-2

gamma = 0.9
theta = 0.000001

V0 = 0.0
V1 = 0.0
iterations = 0

while True:
    new_V0 = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V1)
    new_V1 = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V0)

    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))

    V0 = new_V0
    V1 = new_V1
    iterations += 1

    if delta < theta:
        break

Q_S0_Down = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V1)
Q_S0_Right = -1 + gamma * 0
Q_S1_Right = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V0)
Q_S1_Up = -1 + gamma * V0

print("Assignment 2: Iterative Policy Evaluation")
print("Number of iterations:", iterations)
print(f"V(S0)       = {V0:.6f}")
print(f"V(S1)       = {V1:.6f}")
print(f"Q(S0, Down) = {Q_S0_Down:.6f}")
print(f"Q(S0, Right)= {Q_S0_Right:.6f}")
print(f"Q(S1, Right)= {Q_S1_Right:.6f}")
print(f"Q(S1, Up)   = {Q_S1_Up:.6f}")

print("\nPolicy Consistency Checks")
print(f"V(S0) - Q(S0, Down) = {V0 - Q_S0_Down:.10f}")
print(f"V(S1) - Q(S1, Right) = {V1 - Q_S1_Right:.10f}")
```
