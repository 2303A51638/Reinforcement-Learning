```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines and plays
M = 3
N_play = 10
E = 0.2

# Part B - Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}   # Number of times each machine is selected
R = {'A': 0, 'B': 0, 'C': 0}   # Total reward received from each machine
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}  # Estimated Q-values

total_reward = 0

print("E-Greedy Strategy for Multi-Armed Bandit")
print("------------------------------------------")

# Part C - E-Greedy Decision
for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    if r < E:
        # Exploration
        selected_machine = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        # If multiple machines have same Q-value,
        # randomly select among them
        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number: {r:.4f}")
    print(f"Decision: {decision}")
    print(f"Selected machine: {selected_machine}")

    # Part D - Reward Collection
    reward = float(input(f"Enter reward obtained from machine {selected_machine}: "))

    # Part E - Q-Value Update
    N[selected_machine] += 1
    R[selected_machine] += reward

    # Incremental sample-average formula:
    # Q(a) = Q(a) + [R - Q(a)] / N(a)

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    total_reward += reward

    print(f"Updated Q-value of {selected_machine}: "
          f"{Q[selected_machine]:.2f}")

# Part F - Display Results
print("\n==========================================")
print("FINAL RESULTS")
print("==========================================")

print("\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Value)")

for machine in machines:
    print(
        f"{machine}\t{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t{Q[machine]:.2f}"
    )

print("\nTotal Reward received after 10 plays:", total_reward)

print("\nAnalysis:")
for machine in machines:
    print(
        f"Machine {machine}: selected {N[machine]} times, "
        f"total reward = {R[machine]:.2f}, "
        f"estimated Q-value = {Q[machine]:.2f}"
    )
```
