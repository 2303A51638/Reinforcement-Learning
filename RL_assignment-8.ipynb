```python
gamma = 0.9
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}
transitions = {
    ("S0", "Right"): ("S1", -1),
    ("S1", "Down"): ("S3", -1),
    ("S3", "Right"): ("S4", -1),
    ("S4", "Up"): ("G", 10)
}
def state_value(state):
    total = 0
    discount = 1
    current = state
    while current != "G":
        action = policy[current]
        next_state, reward = transitions[(current, action)]
        total += discount * reward
        discount *= gamma
        current = next_state
    return total
def action_value(state, action):
    if state == "S0" and action == "Up":
        return -1 + gamma * state_value("S0")
    next_state, reward = transitions[(state, action)]
    if next_state == "G":
        return reward
    return reward + gamma * state_value(next_state)
print("State Value of S0:", round(state_value("S0"), 4))
print("Action Value of S0 for Up:", round(action_value("S0", "Up"), 4))
```
